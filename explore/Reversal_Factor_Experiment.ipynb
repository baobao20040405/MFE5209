# Reversal-Factor Experiment

This isolated experiment leaves frozen GMI-ADG unchanged. It replaces only the momentum score with $M^{12-1}+\gamma(-R^{1m})$ for $\gamma\in\{0.5,1.0,2.0\}$ and retains the same EWMA risk model, defensive/offensive rules, next-close execution and 0.05% one-way fee.

```python
from pathlib import Path
import sys
from IPython.display import display, Image
ROOT = Path.cwd()
if not (ROOT / 'reversal_factor_experiment.py').exists():
    ROOT = ROOT / 'explore'
sys.path.insert(0, str(ROOT))
from reversal_factor_experiment import run_experiment, OUTPUT
```

## Results

```python
comparison, correlations = run_experiment()
formatted = comparison.copy()
for column in ['Cumulative Return', 'Annualized Return', 'Annualized Volatility', 'Maximum Drawdown']:
    formatted[column] = formatted[column].map(lambda value: f'{value:.2%}')
for column in ['Sharpe Ratio', 'Sortino Ratio', 'Annual Two-Way Traded Notional']:
    formatted[column] = formatted[column].map(lambda value: f'{value:.2f}')
display(formatted)
display(Image(filename=str(OUTPUT / 'reversal_factor_cumulative_return.png')))
```

## Takeaway

Use the historical-validation rows as the primary decision evidence. A full-history or in-sample improvement alone is insufficient to change the frozen product, particularly when turnover rises.
